Use interactive sliders to watch how the underlying calculations work for a single neuron

first install packages

```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interactive, FloatSlider, VBox
from IPython.display import display
```

define the sigmoid function

```python
def sigmoid(x):
    return 1 / (1 + np.exp(-x))
```

Define the neuron and create the interactive plot with sliders

```python
def neuron(x1, x2, w1, w2, bias):

    # calculate the weighted sum
    z = x1*w1 + x2*w2 + bias

    #calculate neuron output
    y = sigmoid(z)

    #draw sigmoid curve
    xs = np.linspace(-10,10,1000)
    ys = sigmoid(xs)

    plt.figure(figsize=(10,5))
    plt.plot(xs,ys, linewidth =2, label="sigmoid")

    #show current neuron position
    plt.scatter(
        [z],
        [y],
        color="red",
        s=150,
        label=f"Neuron output = {y: .3f}"
    )

    plt.xlabel("Weighted Sum (z)")
    plt.ylabel("Output (y)")
    plt.title("Single Neuron on Sigmoid Curve")
    plt.grid(True)
    plt.legend()

    plt.show()
    print(f"x1 = {x1}")
    print(f"x2 = {x2}")
    print(f"w1 = {w1}")
    print(f"w2 = {w2}")
    print(f"bias = {bias}")
    print()
    print(f"z = {z:.3f}")
    print(f"y = {y:.3f}")

widget = interactive(
    neuron,
    x1=FloatSlider(min = 0, max =1, step = 0.1, value = 1),
    x2=FloatSlider(min = 0, max =1, step = 0.1, value = 0),

    w1=FloatSlider(min = -5, max =5, step = 0.1, value = 2),
    w2=FloatSlider(min = -5, max =5, step = 0.1, value = -1),

    bias=FloatSlider(min = -5, max =5, step = 0.1, value = 0.5)
)

display(widget)

```

```python
from ipywidgets import VBox, interactive_output

def neuron_2(x1,x2, w1, w2, bias):
    plt.close("all")
    # calculate the weighted sum and output

    z = x1*w1 + x2*w2 + bias
    y = sigmoid(z)

    # create figure
    fig, (ax1, ax2) = plt.subplots(
        1,2,
        figsize=(12,5)
    )

    ########################################
    # Left subplot: weighted sum calculation
    ########################################

    ax1.axis("off")

    text = (
        f"Inputs\n"
        f"x1 = {x1:.2f} \n"
        f"x2 = {x2:.2f} \n\n"
        f"Weights\n"
        f"w1 = {w1:.2f} \n"
        f"w2 = {w2:.2f} \n\n"
        f"bias = {bias:.2f} \n"
        f"z = x1*w1 + x2*w2 + bias\n\n"
        f"= ({x1:.2f} x {w1:.2f} x + {x2:.2f} x + {bias:.2f})\n\n"
        f"= {z:.3f} \n\n"
        f"Output = sigmoid(z)\n"
        f"= {y:.3f} \n"
    )

    ax1.text(
        0.05,
        0.5,
        text,
        fontsize=12,
        verticalalignment="center"
    )

    ##############################################
    # Right subplot: sigmoid curve
    ##############################################

    xs = np.linspace(-10,10,1000)
    ys = sigmoid(xs)

    ax2.plot(xs,ys, linewidth =2, label="sigmoid")

    ax2.scatter(
        [z],
        [y],
        color="red",
        s=100,
        zorder=5,
        label=f"Output = {y: .3f}"
    )

    ax2.axvline(
        z,
        color="red",
        linestyle="--",
        alpha=0.6
    )

    ax2.set_title("Single Neuron on Sigmoid Curve")
    ax2.set_xlabel("Weighted Sum (z)")
    ax2.set_ylabel("Output (y)")
    ax2.grid(True)
    ax2.legend()

    ax2.tight_layout()
    plt.show()


controls ={
    "x1" :FloatSlider(value=1, min = 0, max =1, step = 0.1),
    "x2" : FloatSlider(value = 0, min = 0, max =1, step = 0.1),
    "w1": FloatSlider( value = 2, min = -5, max =5, step = 0.1),
    "w2" : FloatSlider(value = -1, min = -5, max =5, step = 0.1),
    "bias" : FloatSlider(value = 0.5, min = -5, max =5, step = 0.1),
}

out = interactive_output(neuron_2, controls)

display(VBox(list(controls.values())))
display(out)
```
