Use interactive sliders to watch how the underlying calculations work for a single neuron

first install packages

```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact, FloatSlider
```

define the sigmoid function

```python
def sigmoid(x):
    return 1 / (1 + np.exp(-x))
```

Define the neuron and create the interactive plot with sliders

```python
def neuron(x1, x2, w1, w2, bias):

    # calculate the weighted sum
    z = x1*w1 + x2*w2 + bias

    #calculate neuron output
    y = sigmoid(z)

    #draw sigmoid curve
    xs = np.linspace(-10,10,1000)
    ys = sigmoid(xs)

    plt.figure(figsize=(10,5))
    plt.plot(xs,ys, linewidth =2, label="sigmoid")

    #show current neuron position
    plt.scatter(
        [z],
        [y],
        color="red",
        s=150,
        label=f"Neuron output = {y: .3f}"
    )

    plt.xlabel("Weighted Sum (z)")
    plt.ylabel("Output (y)")
    plt.title("Single Neuron on Sigmoid Curve")
    plt.grid(True)
    plt.legend()

    plt.show()
    print(f"x1 = {x1}")
    print(f"x2 = {x2}")
    print(f"w1 = {w1}")
    print(f"w2 = {w2}")
    print(f"bias = {bias}")
    print()
    print(f"z = {z:.3f}")
    print(f"y = {y:.3f}")

interact(
    neuron,
    x1=FloatSlider(min = 0, max =1, step = 0.1, value = 1),
    x2=FloatSlider(min = 0, max =1, step = 0.1, value = 0),

    w1=FloatSlider(min = -5, max =5, step = 0.1, value = 2),
    w2=FloatSlider(min = -5, max =5, step = 0.1, value = -1),

    bias=FloatSlider(min = -5, max =5, step = 0.1, value = 0.5)
);

```
